## 1. Setup and Imports

Import necessary modules for UQFF calculations

```python
import numpy as np
import sys
from pathlib import Path

# Physical constants
G = 6.67430e-11          # Gravitational constant (m³/kg·s²)
c = 2.99792458e8         # Speed of light (m/s)
M_sun = 1.98892e30       # Solar mass (kg)
pc = 3.0857e16           # Parsec (m)

print("✅ Constants loaded")
print(f"   G = {G:.3e} m³/kg·s²")
print(f"   c = {c:.3e} m/s")
print(f"   M_sun = {M_sun:.3e} kg")
```

## 2. Example: Sagittarius A* SMBH

Calculate basic properties of Sagittarius A* supermassive black hole

```python
# Sagittarius A* parameters
sagA_params = {
    'name': 'Sagittarius A*',
    'M': 4.1e6 * M_sun,      # 4.1 million solar masses
    'distance': 26000 * pc,  # 26 kpc from Earth
    'type': 'SMBH'
}

print("Sagittarius A* Parameters:")
print(f"  Mass: {sagA_params['M']:.2e} kg ({sagA_params['M']/M_sun:.2e} M_sun)")
print(f"  Distance: {sagA_params['distance']:.2e} m ({sagA_params['distance']/pc:.0f} pc)")
```

```python
# Calculate Schwarzschild radius
r_s = (2 * G * sagA_params['M']) / c**2

print("Schwarzschild Radius:")
print(f"  r_s = 2GM/c²")
print(f"  r_s = {r_s:.3e} m")
print(f"  r_s = {r_s/1e9:.2f} million km")
print(f"  r_s = {r_s/1.496e11:.2f} AU")
```

```python
# Calculate Eddington luminosity
m_p = 1.67262192369e-27  # Proton mass (kg)
sigma_T = 6.6524e-29      # Thomson cross-section (m²)

L_edd = (4 * np.pi * G * sagA_params['M'] * m_p * c) / sigma_T
L_sun = 3.828e26  # Solar luminosity (W)

print("Eddington Luminosity:")
print(f"  L_Edd = (4πGMm_p*c) / σ_T")
print(f"  L_Edd = {L_edd:.3e} W")
print(f"  L_Edd = {L_edd/L_sun:.2e} L_sun")
```

## 3. Galaxy Rotation Curve Example

Calculate rotation velocity for a typical spiral galaxy

```python
# Spiral galaxy parameters (M31 Andromeda-like)
galaxy_params = {
    'name': 'Spiral Galaxy (M31-type)',
    'M': 1.2e12 * M_sun,     # 1.2 trillion solar masses
    'R': 50 * 1000 * pc,     # 50 kpc radius
    'type': 'Spiral Galaxy'
}

print("Galaxy Parameters:")
print(f"  Mass: {galaxy_params['M']:.2e} kg")
print(f"  Radius: {galaxy_params['R']/pc/1e3:.0f} kpc")
```

```python
# Calculate rotation velocity (Newtonian approximation)
r = galaxy_params['R']
M = galaxy_params['M']

v_rot = np.sqrt(G * M / r)

print("Rotation Velocity (Newtonian):")
print(f"  v = sqrt(GM/r)")
print(f"  v = {v_rot:.3e} m/s")
print(f"  v = {v_rot/1e3:.2f} km/s")
print(f"  v/c = {v_rot/c:.2e} (non-relativistic)")
```

## 4. Magnetar Magnetic Field

Calculate magnetic field strength for SGR 1745-2900 magnetar

```python
# SGR 1745-2900 magnetar parameters
magnetar_params = {
    'name': 'SGR 1745-2900',
    'M': 1.4 * M_sun,        # 1.4 solar masses (typical neutron star)
    'R': 12e3,               # 12 km radius
    'B_surface': 1e15,       # 10^15 T surface field
    'type': 'Magnetar'
}

print("Magnetar Parameters:")
print(f"  Mass: {magnetar_params['M']/M_sun:.2f} M_sun")
print(f"  Radius: {magnetar_params['R']/1e3:.0f} km")
print(f"  Surface B-field: {magnetar_params['B_surface']:.1e} T")
```

```python
# Calculate magnetic field at distance r (dipole)
R = magnetar_params['R']
B_surf = magnetar_params['B_surface']
r_test = 100e3  # 100 km from surface

B_r = B_surf * (R / r_test)**3

print(f"Magnetic Dipole Field at r = {r_test/1e3:.0f} km:")
print(f"  B(r) = B_surf * (R/r)³")
print(f"  B({r_test/1e3:.0f} km) = {B_r:.3e} T")

# QED critical field
B_crit = 4.4e13  # Tesla
print(f"\n  B_crit (QED) = {B_crit:.2e} T")
print(f"  B_surface / B_crit = {B_surf/B_crit:.1f}x supercritical")
```
