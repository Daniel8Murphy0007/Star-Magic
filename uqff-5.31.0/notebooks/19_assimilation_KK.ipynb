## Deep-dive: representative observables

For each, we show:
1. **Simple call** — returns `{'value': X}` matching the existing 42-surface contract.
2. **Decomposed call** — adds `decompose=True` to get the 8-field provenance dict
   (value, target, residual_pct, geometry_used, numeric_system, overdetermination_N, alternate_paths, assimilation_status).

### `KK_AU_per_1e10_m`

- **Formula:** `+K^4 - beta - beta^3 - 3`
- **Owner geometry:** `dpm`
- **Documented residual:** 0.5%
- **Primary source:** PAPER_1167

```python
# Simple call — Rule 5 contract (returns {'value': X})
simple = u.calculate_analytic_closures(
    {'qcalcgeom_solve': {'observable': 'KK_AU_per_1e10_m'}})
print('Simple:', simple)
```

```python
# Decomposed — full 8-field provenance dict
decomp = u.calculate_analytic_closures(
    {'qcalcgeom_solve': {'observable': 'KK_AU_per_1e10_m', 'decompose': True}})
for k, v in decomp['value'].items():
    if k != 'alternate_paths':
        print(f'  {k:24s} {v}')
```
