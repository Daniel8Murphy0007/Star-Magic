# UQFF Star-Magic — 5-Minute Quickstart

This notebook walks through UQFF's core surfaces in 5 minutes. Run cell-by-cell.

**Prerequisite:** `pip install uqff` (see `INSTALL.md`)

```python
import uqff_pure_calculator as u

# 34 public calculate_* surfaces
surfaces = sorted(n for n in dir(u) if n.startswith('calculate_'))
print(f'UQFF exposes {len(surfaces)} public surfaces:')
for s in surfaces[:10]:
    print(f'  {s}')
print(f'  ... and {len(surfaces) - 10} more')
```

## 1. Status report — what is this thing?

```python
report = u.calculate_status_report({})['value']['summary']

print(f"Total closures: {report['total_closures']}")
print(f"Truly-independent primitives: {report['truly_independent_primitives']}")
print(f"Derivative primitives: {report['derivative_primitives']}")

print('\nUncertainty classification:')
for cls, n in report['uncertainty_classes_A2_TIER1_production_readiness'].items():
    print(f'  {cls}: {n}')

print('\nCosmic milestones:')
for k, v in report['cosmic_milestones'].items():
    print(f'  {"✅" if v else "⬜"} {k}')
```

## 2. The cosmological constant Λ

UQFF derives Λ from a single primitive (ρ_SCm) and one structural coefficient (K_MEX = 25/12).

```python
ledger = u.calculate_vacuum_ledger({})['value']
for k, v in ledger.items():
    print(f'  {k}: {v}')
```

## 3. The Holmlid 630 eV LENR signature

```python
result = u.calculate_lenr({})['value']
ker = result['ker_chain']
for k, v in ker.items():
    print(f'  {k}: {v}')
```

## 4. The 7 nuclear shell-model magic numbers

All from integer arithmetic on {D_phys=4, SO_5=10, D_crit=26, A_5=60}.

```python
magic = u.calculate_nuclear_magic({})['value']['magic_numbers']
print('Magic number  | UQFF identity from integer primitives')
print('-' * 60)
for k, v in magic.items():
    print(f'  {v:>4}       | {k}')
```

## 5. Resolve a famous physics paradox by name

**Note:** dispatch keys are lowercase — `'hubble_tension'`, not `'Hubble_Tension'`.

```python
import json

for paradox in ['hubble_tension', 'strong_cp_problem', 'cosmological_principle']:
    r = u.calculate_paradox({'paradox': paradox})['value']
    print(f'\n=== {paradox} ===')
    if isinstance(r, dict):
        print(json.dumps(r, indent=2, default=str)[:300])
    else:
        print(r)
```
