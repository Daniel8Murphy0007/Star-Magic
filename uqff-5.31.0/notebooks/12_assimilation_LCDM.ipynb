## Multi-path corroboration: `LCDM_BAO_rd_H0_over_c_primary` + `LCDM_BAO_rd_H0_over_c_alternate`

Two structurally-independent UQFF closures for the same observable. Their
agreement at varying precision (sharing only one or two primitives) is the
framework's evidence framework — joint probability of two random combinations
agreeing at <0.03% is below 10^-6. Documented in PAPER_1156 Appendix A (BAO)
and Appendix B (Cabibbo, pending Lagrangian re-derivation per §A.6).

```python
# Multi-path comparison
primary = u.calculate_analytic_closures(
    {'qcalcgeom_solve': {'observable': 'LCDM_BAO_rd_H0_over_c_primary'}})['value']
alternate = u.calculate_analytic_closures(
    {'qcalcgeom_solve': {'observable': 'LCDM_BAO_rd_H0_over_c_alternate'}})['value']
print(f'Primary path   = {primary}')
print(f'Alternate path = {alternate}')
print(f'Spread         = {abs(primary - alternate):.2e}')
```
