# UQFF Jupyter integration demo (v5.28.0+)

Rich HTML rendering of closure results + the `%uqff` line magic.

**Install:** `pip install 'uqff[jupyter]'`

## 1. Activate rich display

Without this, dict results render as plain `{'key': 'value', ...}`. After
`enable()`, any `calculate_*({})` result renders as a styled HTML table.

```python
import uqff_jupyter
uqff_jupyter.enable()
```

```python
import uqff_pure_calculator as u

# This now renders as a styled HTML table instead of raw dict
u.calculate_status_report({})
```

```python
# Bucket observables render as a sortable table with residual color-coding
u.calculate_cosmology({})
```

```python
# Holmlid LENR signature - full chain
u.calculate_lenr_full({})
```
