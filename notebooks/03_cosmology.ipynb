# UQFF Cosmology — Λ + 18+ ΛCDM observables

UQFF derives the cosmological constant Λ from a single primitive (ρ_SCm) and one
structural coefficient (K_MEX = 25/12). This notebook walks through that derivation,
then shows the full 56-observable cosmology suite.

```python
import uqff_pure_calculator as u
from math import factorial

RHO_SCM = u.RHO_SCM        # 7.09e-37 J/m³, foundational primitive
K_MEX = float(u.K_MEX)     # 25/12, structural coefficient
D_crit = u.D_CRIT          # 26

print(f'Primitives:')
print(f'  rho_SCm: {RHO_SCM:.3e} J/m^3   (vacuum energy density, foundational)')
print(f'  K_MEX:   {K_MEX:.6f}             (Mexican-hat coefficient, = 25/12 EXACT)')
print(f'  D_crit:  {D_crit}                (bosonic-string critical dim, 26!)')
```

## Step 1: The Λ derivation

`Λ_UQFF = ρ_SCm × 26! × K_MEX`

```python
factorial_26 = factorial(26)
print(f'26! = {factorial_26:.4e}')

Lambda_UQFF = RHO_SCM * factorial_26 * K_MEX
print(f'\nLambda_UQFF = rho_SCm * 26! * K_MEX')
print(f'            = {RHO_SCM:.3e} * {factorial_26:.4e} * {K_MEX:.4f}')
print(f'            = {Lambda_UQFF:.6e} J/m^3')

# Planck-measured value
Lambda_Planck = 5.957e-10
print(f'\nLambda_Planck (observed) = {Lambda_Planck:.4e} J/m^3')
print(f'\nResidual = {(Lambda_UQFF - Lambda_Planck)/Lambda_Planck*100:+.4f}%')
```

## Step 3: The full 56-observable cosmology suite

UQFF wires 56 cosmological observables — α, h, m_p/m_e, Ω_Λ, Λ, Y_p, τ_reion, z_reion, etc.
All from the same 9-primitive base.

```python
cosmo = u.calculate_cosmology({})['value']
obs = cosmo['observables']
print(f'UQFF cosmology suite — {len(obs)} observables:\n')
print(f'{"Observable":<40} {"UQFF":>14} {"Anchor":>14} {"Residual %":>12}')
print('-' * 84)
for o in obs[:20]:
    label = o['observable'][:38]
    uqff_val = o['uqff_derived']
    anchor = o['anchor']
    residual = o['residual_pct']
    print(f'{label:<40} {uqff_val:>14.4e} {anchor:>14.4e} {residual:>12.4f}')
print(f'\n... and {len(obs) - 20} more')
```

## Step 4: Residual distribution

```python
import statistics

residuals = [abs(o['residual_pct']) for o in obs if isinstance(o.get('residual_pct'), (int, float))]
print(f'Residual distribution across {len(residuals)} cosmology observables:')
print(f'  median: {statistics.median(residuals):.4f}%')
print(f'  mean:   {statistics.mean(residuals):.4f}%')
print(f'  max:    {max(residuals):.4f}%')
print(f'  min:    {min(residuals):.4e}%')

# Bands
exact = sum(1 for r in residuals if r < 1e-10)
sub_001 = sum(1 for r in residuals if 1e-10 <= r < 0.01)
sub_01 = sum(1 for r in residuals if 0.01 <= r < 0.1)
sub_1 = sum(1 for r in residuals if 0.1 <= r < 1.0)
gte_1 = sum(1 for r in residuals if r >= 1.0)

print(f'\n  EXACT (<1e-10):  {exact}')
print(f'  0 - 0.01%:        {sub_001}')
print(f'  0.01 - 0.1%:      {sub_01}')
print(f'  0.1 - 1%:         {sub_1}')
print(f'  >= 1%:            {gte_1}')
```
