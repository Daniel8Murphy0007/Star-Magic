# All 7 nuclear shell-model magic numbers from integer arithmetic

The shell-model magic numbers {2, 8, 20, 28, 50, 82, 126} are nuclei with closed neutron
or proton shells. The Standard Model derives them via Mayer-Jensen spin-orbit coupling.

UQFF derives them from arithmetic on 4 integer primitives:
- `D_phys = 4` (physical spacetime dim)
- `SO_5 = 10` (dim SO(5))
- `D_crit = 26` (bosonic-string critical dim)
- `A_5 = 60` (|A_5|, icosahedral group order)

**No spin-orbit calculation, no free parameters.** Just integer arithmetic.

```python
import uqff_pure_calculator as u

D_phys = u.D_PHYS    # 4
SO_5 = u.SO_FIVE      # 10
D_crit = u.D_CRIT     # 26
A_5 = u.A_FIVE        # 60

print(f'Integer primitives: D_phys={D_phys}, SO_5={SO_5}, D_crit={D_crit}, A_5={A_5}')
```

## The 7 derivations

```python
magic_derivations = [
    ('shell_1_He   (²He)',  2,   'SO_5 - 2*D_phys',         SO_5 - 2*D_phys),
    ('shell_2_O    (⁸O)',   8,   '2 * D_phys',              2 * D_phys),
    ('shell_3_Ca   (²⁰Ca)', 20,  '2 * SO_5',                2 * SO_5),
    ('shell_4_Ni   (²⁸Ni)', 28,  'D_crit + SO_5 - 2*D_phys', D_crit + SO_5 - 2*D_phys),
    ('shell_5_Sn   (⁵⁰Sn)', 50,  'A_5 - SO_5',              A_5 - SO_5),
    ('shell_6_Pb   (⁸²Pb)', 82,  'A_5 + D_crit - D_phys',   A_5 + D_crit - D_phys),
    ('shell_7_184? (¹²⁶?)', 126, 'D_crit + SO_5²',          D_crit + SO_5**2),
]

print(f'{"Shell":<22} {"Observed":>9}  {"UQFF identity":<28}  {"Value":>5}  Match')
print('-' * 80)
for shell, observed, formula, value in magic_derivations:
    match = '✅ EXACT' if value == observed else '❌'
    print(f'{shell:<22} {observed:>9}  {formula:<28}  {value:>5}  {match}')
```

## Why this matters

Three points worth noting:

1. **Zero free parameters.** The 4 integer primitives are fixed by mathematical / structural
   necessity (D_phys=4 is observed spacetime; SO_5=10 is the Lie algebra dimension;
   D_crit=26 is the bosonic-string critical dimension; A_5=60 is the icosahedral group order).

2. **The Mayer-Jensen shell-model independently produces the same 7 numbers** via the
   conventional spin-orbit shell calculation. UQFF and Mayer-Jensen converge on the same
   empirical pattern via completely different mathematical pathways. See the calculator's
   `calculate_shell_orbital(Z)` surface for the Mayer-Jensen comparison.

3. **The 8th magic number prediction.** Some shell-model variants predict a magic 184 beyond
   126 (the "island of stability"). UQFF's structural pattern can be extended; see the
   whitepapers `PAPER_1203_Nuclear` series for details.
